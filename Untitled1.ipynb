```python
from google.colab import drive
import os, joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# 1. Kết nối Drive và trỏ tới thư mục dataset mới đã tiền xử lý
drive.mount('/content/drive')
DRIVE_PROJECT = '/content/drive/MyDrive/Parameter_Identification_of_PMSM_using_PINN'
PROCESSED_DATA_DIR = os.path.join(DRIVE_PROJECT, 'dataset', 'processed')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Thiết bị tính toán:", device)

X_scaler = joblib.load(os.path.join(PROCESSED_DATA_DIR, 'X_scaler.pkl'))
Y_scaler = joblib.load(os.path.join(PROCESSED_DATA_DIR, 'Y_scaler.pkl'))

X_train = torch.tensor(np.load(os.path.join(PROCESSED_DATA_DIR, 'X_train.npy')), dtype=torch.float32, device=device)
Y_train_ldq = torch.tensor(np.load(os.path.join(PROCESSED_DATA_DIR, 'Y_train.npy'))[:, 1:3], dtype=torch.float32, device=device)

X_val = torch.tensor(np.load(os.path.join(PROCESSED_DATA_DIR, 'X_validation.npy')), dtype=torch.float32, device=device)
Y_val_ldq = torch.tensor(np.load(os.path.join(PROCESSED_DATA_DIR, 'Y_validation.npy'))[:, 1:3], dtype=torch.float32, device=device)

X_test = torch.tensor(np.load(os.path.join(PROCESSED_DATA_DIR, 'X_test.npy')), dtype=torch.float32, device=device)
Y_test_s = np.load(os.path.join(PROCESSED_DATA_DIR, 'Y_test.npy'))
T_test = np.load(os.path.join(PROCESSED_DATA_DIR, 'Time_test.npy')).flatten()
S_test = np.load(os.path.join(PROCESSED_DATA_DIR, 'Steady_test.npy')) > 0.5

# 2. Huấn luyện mạng MLP 2 đầu ra [Ld, Lq]
class MLP_Inductance(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(5, 128), nn.Tanh(),
            nn.Linear(128, 128), nn.Tanh(),
            nn.Linear(128, 128), nn.Tanh(),
            nn.Linear(128, 2)
        )
    def forward(self, x):
        return self.net(x)

model_ldq = MLP_Inductance().to(device)
optimizer = optim.Adam(model_ldq.parameters(), lr=0.002)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=15)
criterion = nn.MSELoss()

best_val = float('inf')
for epoch in range(1, 151):
    model_ldq.train()
    perm = torch.randperm(len(X_train), device=device)
    for i in range(0, len(X_train), 512):
        idx = perm[i:i+512]
        loss = criterion(model_ldq(X_train[idx]), Y_train_ldq[idx])
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    model_ldq.eval()
    with torch.no_grad():
        val_loss = criterion(model_ldq(X_val), Y_val_ldq).item()
    scheduler.step(val_loss)
    if val_loss < best_val:
        best_val = val_loss
        torch.save(model_ldq.state_dict(), '/content/best_mlp_ldq.pth')

# Dự đoán Ld, Lq trên tập Test
model_ldq.load_state_dict(torch.load('/content/best_mlp_ldq.pth'))
model_ldq.eval()
with torch.no_grad():
    pred_ldq_s = model_ldq(X_test).cpu().numpy()

Ld_hat = pred_ldq_s[:, 0] * Y_scaler.scale_[1] + Y_scaler.mean_[1]
Lq_hat = pred_ldq_s[:, 1] * Y_scaler.scale_[2] + Y_scaler.mean_[2]
print("Hoàn tất Giai đoạn 1: Đã nhận dạng xong Ld, Lq!")
```

```python
# ===== CELL 2: SỬA LỖI CỬA SỔ TÍCH PHÂN (CHỈ TÍCH PHÂN KHI CÓ TẢI & KHỬ LỆCH BƯỚC NHẢY) =====
X_test_r = X_test.cpu().numpy() * X_scaler.scale_ + X_scaler.mean_
Y_test_r = Y_scaler.inverse_transform(Y_test_s)

ud, uq, Id, Iq, nm = X_test_r[:, 0], X_test_r[:, 1], X_test_r[:, 2], X_test_r[:, 3], X_test_r[:, 4]
Rs_true, Ld_true, Lq_true = Y_test_r[:, 0], Y_test_r[:, 1], Y_test_r[:, 2]

omega_e = 9.0 * nm * (2.0 * np.pi / 60.0)

# 1. Tính tử số a_k và mẫu số b_k chuẩn theo phương trình trục q
# Chỉ dùng các bước liên tục (dt < 1e-4s) để tránh sai phân giả khi file 02 cắt bỏ mẫu trùng
dt = np.diff(T_test, prepend=T_test[0])
dt[0] = 5e-5
continuous = dt <= 1.5e-4

diq = np.diff(Iq, prepend=Iq[0])
diq[~continuous] = 0.0
dt_clean = np.where(continuous, dt, 5e-5)

a_k = (uq - omega_e * (Ld_hat * Id + 0.2319)) * dt_clean - Lq_hat * diq
b_k = Iq * dt_clean

# 2. Chỉ tích phân trên các mẫu có dòng tải đáng kể (|Iq| > 10 A) để cửa sổ không bị nhiễm rác lúc không tải
load_mask = np.abs(Iq) > 10.0
a_load = np.where(load_mask, a_k, 0.0)
b_load = np.where(load_mask, b_k, 0.0)

W = 300
sum_a = pd.Series(a_load).rolling(W, min_periods=1).sum().to_numpy()
sum_b = pd.Series(b_load).rolling(W, min_periods=1).sum().to_numpy()

Rs_hat = np.full_like(Rs_true, np.nan)
valid = np.abs(sum_b) > 0.05
Rs_hat[valid] = sum_a[valid] / sum_b[valid]
Rs_hat = np.clip(Rs_hat, 0.033, 0.046)
Rs_hat = pd.Series(Rs_hat).ffill().bfill().to_numpy()

# 3. Kiểm tra sai số và độ lệch điện áp chuẩn của dataset
mape = lambda yt, yp: np.mean(np.abs((yt - yp) / yt)) * 100.0
print(f"Sai số Rs (Toàn bộ): {mape(Rs_true, Rs_hat):.2f}%")
print(f"Sai số Ld (Toàn bộ): {mape(Ld_true, Ld_hat):.2f}%")
print(f"Sai số Lq (Toàn bộ): {mape(Lq_true, Lq_hat):.2f}%")

# Chẩn đoán nhanh: Kiểm tra phương trình trục q với chính nhãn thật (Rs_true, Ld_true)
du_check = uq[load_mask] - (Rs_true[load_mask] * Iq[load_mask] + omega_e[load_mask] * (Ld_true[load_mask] * Id[load_mask] + 0.2319))
print(f"Độ lệch trung bình phương trình uq trên tập Test: {np.mean(du_check):.4f} V")
```
